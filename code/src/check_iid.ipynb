```python
from cleanlab import Datalab

from service.datasetservice.DatasetService import DatasetService

dataset_service = DatasetService()
x_all, y_all = dataset_service.get_subject_data(subject="all", with_features=True)


# features and (optional) labels are numpy arrays
dataset = {'features': x_all.to_numpy(), 'labels': y_all.to_numpy()}
datalab = Datalab(dataset, label_name='labels')

# tell Datalab that we only want to run its non-IID check
datalab.find_issues(features=dataset['features'], issue_types={'non_iid': {}})

p_value = datalab.get_info('non_iid')['p-value']  # low value suggests data are not IID
```
