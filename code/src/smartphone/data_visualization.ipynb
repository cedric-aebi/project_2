```python
import pandas as pd

df = pd.read_csv("../../dataset/stresSense.csv")
# Filter data for user 1 and eating activity
user_1_data = df[(df['User'] == 1)]
print(user_1_data.shape)
user_1_data.head()
```

```python
import matplotlib.pyplot as plt

# Filter data for user 1 and eating activity
data_for_plot = user_1_data[user_1_data['Activity'] == "Eating"]

# Get data from 250 to 500 by index
user_1_data_first_5_seconds = data_for_plot.iloc[200:700]

# Plot the accelerometer data
plt.figure(figsize=(10, 6))
plt.plot(user_1_data_first_5_seconds.index, user_1_data_first_5_seconds['Ax'], label='Ax')
plt.plot(user_1_data_first_5_seconds.index, user_1_data_first_5_seconds['Ay'], label='Ay')
plt.plot(user_1_data_first_5_seconds.index, user_1_data_first_5_seconds['Az'], label='Az')
plt.title('Accelerometer Data for User 1 (First 5 Seconds)')
plt.xlabel('Time')
plt.ylabel('Acceleration')
plt.legend()
plt.grid(True)
plt.show()
```
